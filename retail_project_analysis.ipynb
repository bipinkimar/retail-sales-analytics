```python
## Preserve raw data
customer_clean = customer.copy()
order_clean = order.copy()
order_item_clean = order_item.copy()
product_clean = product.copy()
returns_clean = returns.copy()
marketing_clean = marketing.copy()
```

# Data Claening

```python
## check the clean data counts
customer_before = len(customer_clean)
order_before = len(order_clean)
order_item_before = len(order_item_clean)
```

```python
## remove exact duplicates
customer_clean = customer_clean.drop_duplicates()
order_clean = order_clean.drop_duplicates()
order_item_clean = order_item_clean.drop_duplicates()
```

```python
## Standardize categorical values

## for customer city
customer_clean["city"] = customer_clean["city"].replace({"delhi":"Delhi"})
```

```python
## for order channel
order_clean["channel"] = order_clean["channel"].replace({"online":"Online"})
```

```python
## for product
product_clean["category"] = product_clean["category"].replace({"electronics":"Electronics"})
```

```python
product_clean["category"].value_counts(dropna=False)
```

```python
## remove unwanted spaces
## for customer

customer_clean["gender"] = customer_clean["gender"].str.strip()
customer_clean["city"] = customer_clean["city"].str.strip()
customer_clean["state"] = customer_clean["state"].str.strip()
customer_clean["customer_segment"] = customer_clean["customer_segment"].str.strip()
```

```python
## for order
order_clean["channel"] = order_clean["channel"].str.strip()
order_clean["payment_method"] = order_clean["payment_method"].str.strip()
order_clean["order_status"] = order_clean["order_status"].str.strip()
```

```python
## for product
product_clean["product_name"] = product_clean["product_name"].str.strip()
product_clean["category"] = product_clean["category"].str.strip()
```

```python
## missing-value treatment.
## check whose are fit mean vs meadian

print(customer_clean["age"].describe())
```

```python
## check distribution of age
customer_clean["age"].value_counts().sort_index()
```

```python
## age convert into integer data type
customer_clean["age"] = customer_clean["age"].astype(int)
```

```python
## fill age with median
customer_clean["age"] = customer_clean["age"].fillna(customer_clean["age"].median())
```

```python
## check the missing values
customer_clean["age"].isnull().sum()
```

```python
## check the customer segment

customer_clean["customer_segment"].value_counts(dropna=False)
```

```python
## fill with the unknown values
customer_clean["customer_segment"] = customer_clean["customer_segment"].fillna("Unknown")
```

```python
## verify the data
customer_clean["customer_segment"].value_counts(dropna=False)
```

```python
customer_clean["customer_segment"].isnull().sum()
```

```python
## for payment method fill with unknown data
order_clean["payment_method"] = order_clean["payment_method"].fillna("Unknown")
```

```python
## verify
print(order_clean["payment_method"].value_counts(dropna=False))
print(order_clean["payment_method"].isnull().sum())
```

```python
## check marketing spend
print(order_item_clean["marketing_spend"].describe())
print(order_item_clean["marketing_spend"].isnull().sum())
```

```python
## for return reason fill with unknown values
returns_clean["return_reason"] = returns_clean["return_reason"].fillna("Unknown")
```

```python
## verify the return reason
returns_clean["return_reason"].value_counts(dropna=False)
```
